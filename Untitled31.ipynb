```python
# ============================================================
# TASK 11 - MATCHING & RANKING V2
# PYTHON ANALYSIS ONLY
# Charts will be created later in Tableau
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display

# ============================================================
# 1. LOAD CLEANED DATASET
# ============================================================

file_path = "/content/Cleaned_Matching_Ranking_Dataset.xlsx"

df = pd.read_excel(file_path)

print("=" * 70)
print("TASK 11 - MATCHING & RANKING ANALYSIS")
print("=" * 70)

print("\nDataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())


# ============================================================
# 2. DATASET OVERVIEW
# ============================================================

print("\n" + "=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

total_matches = len(df)
total_jobs = df["Job_ID"].nunique()
total_candidates = df["Candidate_ID"].nunique()
total_categories = df["Job_Category"].nunique()

print("Total Matches     :", total_matches)
print("Unique Jobs       :", total_jobs)
print("Unique Candidates :", total_candidates)
print("Job Categories    :", total_categories)


# ============================================================
# 3. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 70)
print("DATA QUALITY CHECK")
print("=" * 70)

print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing_Count"))

print("\nDuplicate Records:", df.duplicated().sum())

print("\nData Types:")
display(df.dtypes.to_frame("Data_Type"))


# ============================================================
# 4. BASIC STATISTICAL ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STATISTICAL ANALYSIS")
print("=" * 70)

numeric_columns = [
    "Experience_Years",
    "Skills_Match_Pct",
    "Profile_Quality_Score",
    "Relevance_Score",
    "Time_to_Hire_Days",
    "Rank_Position",
    "User_Rating",
    "Baseline_Score",
    "New_Ranker_Score"
]

statistics = df[numeric_columns].describe().T

statistics["median"] = df[numeric_columns].median()
statistics["variance"] = df[numeric_columns].var()
statistics["std_dev"] = df[numeric_columns].std()

display(statistics.round(3))


# ============================================================
# 5. CORE MATCH QUALITY METRICS
# ============================================================

print("\n" + "=" * 70)
print("CORE MATCH QUALITY METRICS")
print("=" * 70)

shortlisted = (df["Shortlisted"] == "Yes").sum()
interviewed = (df["Interviewed"] == "Yes").sum()
hired = (df["Hired"] == "Yes").sum()

shortlist_rate = shortlisted / total_matches * 100
interview_rate = interviewed / total_matches * 100
hire_rate = hired / total_matches * 100

avg_relevance = df["Relevance_Score"].mean()
avg_time_to_hire = df["Time_to_Hire_Days"].mean()
avg_skills_match = df["Skills_Match_Pct"].mean()
avg_profile_quality = df["Profile_Quality_Score"].mean()
avg_rating = df["User_Rating"].mean()

print(f"Total Matches          : {total_matches}")
print(f"Average Relevance      : {avg_relevance:.4f}")
print(f"Shortlist Rate         : {shortlist_rate:.2f}%")
print(f"Interview Rate         : {interview_rate:.2f}%")
print(f"Hire Rate              : {hire_rate:.2f}%")
print(f"Average Time-to-Hire   : {avg_time_to_hire:.2f} days")
print(f"Average Skills Match   : {avg_skills_match:.2f}%")
print(f"Average Profile Score  : {avg_profile_quality:.2f}")
print(f"Average User Rating    : {avg_rating:.2f}")


# ============================================================
# 6. METRICS TABLE
# ============================================================

metrics = pd.DataFrame({
    "Metric": [
        "Total Matches",
        "Unique Jobs",
        "Unique Candidates",
        "Average Relevance",
        "Shortlist Rate",
        "Interview Rate",
        "Hire Rate",
        "Average Time-to-Hire",
        "Average Skills Match",
        "Average Profile Quality",
        "Average User Rating"
    ],
    "Value": [
        total_matches,
        total_jobs,
        total_candidates,
        avg_relevance,
        shortlist_rate,
        interview_rate,
        hire_rate,
        avg_time_to_hire,
        avg_skills_match,
        avg_profile_quality,
        avg_rating
    ]
})

print("\nMetrics Summary:")
display(metrics.round(3))


# ============================================================
# 7. CATEGORY-WISE ANALYSIS
# ============================================================

category_analysis = df.groupby("Job_Category").agg(
    Total_Matches=("Match_ID", "count"),
    Avg_Relevance=("Relevance_Score", "mean"),
    Avg_Skills_Match=("Skills_Match_Pct", "mean"),
    Avg_Profile_Quality=("Profile_Quality_Score", "mean"),
    Shortlisted=("Shortlisted", lambda x: (x == "Yes").sum()),
    Interviewed=("Interviewed", lambda x: (x == "Yes").sum()),
    Hired=("Hired", lambda x: (x == "Yes").sum()),
    Avg_Time_to_Hire=("Time_to_Hire_Days", "mean"),
    Avg_Rating=("User_Rating", "mean")
).reset_index()

category_analysis["Shortlist_Rate"] = (
    category_analysis["Shortlisted"] /
    category_analysis["Total_Matches"] * 100
)

category_analysis["Interview_Rate"] = (
    category_analysis["Interviewed"] /
    category_analysis["Total_Matches"] * 100
)

category_analysis["Hire_Rate"] = (
    category_analysis["Hired"] /
    category_analysis["Total_Matches"] * 100
)

category_analysis = category_analysis.round(3)

print("\n" + "=" * 70)
print("CATEGORY-WISE PERFORMANCE")
print("=" * 70)

display(category_analysis)


# ============================================================
# 8. BEST & WORST CATEGORY
# ============================================================

best_relevance_category = category_analysis.loc[
    category_analysis["Avg_Relevance"].idxmax(),
    "Job_Category"
]

best_shortlist_category = category_analysis.loc[
    category_analysis["Shortlist_Rate"].idxmax(),
    "Job_Category"
]

best_hire_category = category_analysis.loc[
    category_analysis["Hire_Rate"].idxmax(),
    "Job_Category"
]

fastest_hire_category = category_analysis.loc[
    category_analysis["Avg_Time_to_Hire"].idxmin(),
    "Job_Category"
]

print("\n" + "=" * 70)
print("CATEGORY INSIGHTS")
print("=" * 70)

print("Highest Average Relevance :", best_relevance_category)
print("Highest Shortlist Rate    :", best_shortlist_category)
print("Highest Hire Rate         :", best_hire_category)
print("Fastest Time-to-Hire      :", fastest_hire_category)


# ============================================================
# 9. EVALUATION SET ANALYSIS
# ============================================================

evaluation_analysis = (
    df.groupby("Evaluation_Label")
    .agg(
        Matches=("Match_ID", "count"),
        Avg_Relevance=("Relevance_Score", "mean"),
        Avg_New_Ranker=("New_Ranker_Score", "mean"),
        Avg_Baseline=("Baseline_Score", "mean"),
        Shortlisted=("Shortlisted", lambda x: (x == "Yes").sum()),
        Hired=("Hired", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

evaluation_analysis["Percentage"] = (
    evaluation_analysis["Matches"] /
    evaluation_analysis["Matches"].sum() * 100
)

evaluation_analysis = evaluation_analysis.round(3)

print("\n" + "=" * 70)
print("LABELLED EVALUATION SET")
print("=" * 70)

display(evaluation_analysis)


# ============================================================
# 10. RELEVANCE QUALITY GROUPS
# ============================================================

df["Relevance_Group"] = pd.cut(
    df["Relevance_Score"],
    bins=[0, 0.40, 0.60, 0.80, 1.00],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ],
    include_lowest=True
)

relevance_analysis = df.groupby(
    "Relevance_Group",
    observed=False
).agg(
    Matches=("Match_ID", "count"),
    Avg_Skills_Match=("Skills_Match_Pct", "mean"),
    Avg_Profile_Quality=("Profile_Quality_Score", "mean"),
    Shortlist_Rate=("Shortlisted", lambda x: (x == "Yes").mean() * 100),
    Hire_Rate=("Hired", lambda x: (x == "Yes").mean() * 100)
).reset_index()

print("\n" + "=" * 70)
print("RELEVANCE QUALITY ANALYSIS")
print("=" * 70)

display(relevance_analysis.round(3))


# ============================================================
# 11. SHORTLIST PERFORMANCE
# ============================================================

shortlist_analysis = df.groupby("Shortlisted").agg(
    Total_Matches=("Match_ID", "count"),
    Avg_Relevance=("Relevance_Score", "mean"),
    Avg_Skills_Match=("Skills_Match_Pct", "mean"),
    Avg_Profile_Quality=("Profile_Quality_Score", "mean"),
    Avg_Ranker_Score=("New_Ranker_Score", "mean")
).reset_index()

print("\n" + "=" * 70)
print("SHORTLIST PERFORMANCE")
print("=" * 70)

display(shortlist_analysis.round(3))


# ============================================================
# 12. HIRING PERFORMANCE
# ============================================================

hiring_analysis = df.groupby("Hired").agg(
    Total_Matches=("Match_ID", "count"),
    Avg_Relevance=("Relevance_Score", "mean"),
    Avg_Skills_Match=("Skills_Match_Pct", "mean"),
    Avg_Profile_Quality=("Profile_Quality_Score", "mean"),
    Avg_Time_to_Hire=("Time_to_Hire_Days", "mean"),
    Avg_Rating=("User_Rating", "mean"),
    Avg_New_Ranker=("New_Ranker_Score", "mean")
).reset_index()

print("\n" + "=" * 70)
print("HIRING PERFORMANCE")
print("=" * 70)

display(hiring_analysis.round(3))


# ============================================================
# 13. BASELINE VS NEW RANKER
# ============================================================

baseline_avg = df["Baseline_Score"].mean()
new_ranker_avg = df["New_Ranker_Score"].mean()

absolute_improvement = new_ranker_avg - baseline_avg

percentage_improvement = (
    absolute_improvement / baseline_avg
) * 100

print("\n" + "=" * 70)
print("BASELINE VS NEW RANKER")
print("=" * 70)

print(f"Baseline Average Score : {baseline_avg:.4f}")
print(f"New Ranker Average     : {new_ranker_avg:.4f}")
print(f"Absolute Improvement   : {absolute_improvement:.4f}")
print(f"Improvement Percentage  : {percentage_improvement:.2f}%")


# ============================================================
# 14. RANKER IMPROVEMENT BY CATEGORY
# ============================================================

ranker_category = df.groupby("Job_Category").agg(
    Baseline_Score=("Baseline_Score", "mean"),
    New_Ranker_Score=("New_Ranker_Score", "mean")
).reset_index()

ranker_category["Improvement"] = (
    ranker_category["New_Ranker_Score"] -
    ranker_category["Baseline_Score"]
)

ranker_category["Improvement_Pct"] = (
    ranker_category["Improvement"] /
    ranker_category["Baseline_Score"] * 100
)

print("\n" + "=" * 70)
print("RANKER IMPROVEMENT BY CATEGORY")
print("=" * 70)

display(ranker_category.round(4))


# ============================================================
# 15. TOP MATCHES
# ============================================================

top_matches = df.sort_values(
    "New_Ranker_Score",
    ascending=False
).head(10)

print("\n" + "=" * 70)
print("TOP 10 MATCHES")
print("=" * 70)

display(
    top_matches[
        [
            "Match_ID",
            "Job_ID",
            "Candidate_ID",
            "Job_Category",
            "Skills_Match_Pct",
            "Relevance_Score",
            "New_Ranker_Score",
            "Evaluation_Label",
            "Shortlisted",
            "Hired"
        ]
    ]
)


# ============================================================
# 16. TOP 10% RANKER PERFORMANCE
# ============================================================

top_10_percent = df["New_Ranker_Score"].quantile(0.90)

top_ranked = df[
    df["New_Ranker_Score"] >= top_10_percent
]

top_10_relevance = top_ranked["Relevance_Score"].mean()
top_10_shortlist_rate = (
    top_ranked["Shortlisted"] == "Yes"
).mean() * 100

top_10_hire_rate = (
    top_ranked["Hired"] == "Yes"
).mean() * 100

print("\n" + "=" * 70)
print("TOP 10% RANKER PERFORMANCE")
print("=" * 70)

print(f"Top 10% Score Threshold : {top_10_percent:.4f}")
print(f"Average Relevance       : {top_10_relevance:.4f}")
print(f"Shortlist Rate          : {top_10_shortlist_rate:.2f}%")
print(f"Hire Rate               : {top_10_hire_rate:.2f}%")


# ============================================================
# 17. RANK POSITION PERFORMANCE
# ============================================================

rank_analysis = df.groupby("Rank_Position").agg(
    Matches=("Match_ID", "count"),
    Avg_Relevance=("Relevance_Score", "mean"),
    Avg_Baseline=("Baseline_Score", "mean"),
    Avg_New_Ranker=("New_Ranker_Score", "mean"),
    Shortlist_Rate=("Shortlisted", lambda x: (x == "Yes").mean() * 100),
    Hire_Rate=("Hired", lambda x: (x == "Yes").mean() * 100)
).reset_index()

print("\n" + "=" * 70)
print("RANK POSITION PERFORMANCE")
print("=" * 70)

display(rank_analysis.round(3))


# ============================================================
# 18. CORRELATION ANALYSIS
# ============================================================

correlation_columns = [
    "Experience_Years",
    "Skills_Match_Pct",
    "Profile_Quality_Score",
    "Relevance_Score",
    "Time_to_Hire_Days",
    "Rank_Position",
    "User_Rating",
    "Baseline_Score",
    "New_Ranker_Score"
]

correlation_matrix = df[correlation_columns].corr()

print("\n" + "=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)

display(correlation_matrix.round(3))


# ============================================================
# 19. STRONGEST RELATIONSHIPS
# ============================================================

corr_pairs = (
    correlation_matrix
    .where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)

corr_pairs.columns = [
    "Variable_1",
    "Variable_2",
    "Correlation"
]

corr_pairs["Absolute_Correlation"] = (
    corr_pairs["Correlation"].abs()
)

strongest_relationships = corr_pairs.sort_values(
    "Absolute_Correlation",
    ascending=False
).head(10)

print("\n" + "=" * 70)
print("STRONGEST VARIABLE RELATIONSHIPS")
print("=" * 70)

display(strongest_relationships.round(3))


# ============================================================
# 20. BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("KEY BUSINESS INSIGHTS")
print("=" * 70)

print(f"""
1. Matching Volume
   The dataset contains {total_matches} match records across
   {total_jobs} jobs and {total_candidates} candidates.

2. Relevance
   The average relevance score is {avg_relevance:.4f},
   indicating the overall matching relevance level.

3. Shortlisting
   The overall shortlist rate is {shortlist_rate:.2f}%,
   showing the percentage of matches progressing to shortlist.

4. Hiring
   The overall hire rate is {hire_rate:.2f}%,
   showing the percentage of matches resulting in hiring.

5. Time-to-Hire
   Average time-to-hire is {avg_time_to_hire:.2f} days.

6. Skills Alignment
   Average skills match is {avg_skills_match:.2f}%.

7. Ranking Performance
   Baseline average score is {baseline_avg:.4f},
   while the new ranker average score is {new_ranker_avg:.4f}.

8. Ranker Improvement
   The new ranker shows an average score improvement of
   {percentage_improvement:.2f}% over the baseline.

9. Best Relevance Category
   {best_relevance_category} has the highest average relevance.

10. Best Shortlist Category
    {best_shortlist_category} has the highest shortlist rate.

11. Best Hire Category
    {best_hire_category} has the highest hire rate.

12. Fastest Hiring Category
    {fastest_hire_category} has the lowest average time-to-hire.
""")


# ============================================================
# 21. FINAL KPI TABLE FOR TABLEAU / REPORT
# ============================================================

final_kpis = pd.DataFrame({
    "KPI": [
        "Total Matches",
        "Unique Jobs",
        "Unique Candidates",
        "Average Relevance",
        "Shortlist Rate",
        "Interview Rate",
        "Hire Rate",
        "Average Time-to-Hire",
        "Average Skills Match",
        "Average Profile Quality",
        "Average User Rating",
        "Baseline Score",
        "New Ranker Score",
        "Ranker Improvement %"
    ],
    "Value": [
        total_matches,
        total_jobs,
        total_candidates,
        round(avg_relevance, 4),
        round(shortlist_rate, 2),
        round(interview_rate, 2),
        round(hire_rate, 2),
        round(avg_time_to_hire, 2),
        round(avg_skills_match, 2),
        round(avg_profile_quality, 2),
        round(avg_rating, 2),
        round(baseline_avg, 4),
        round(new_ranker_avg, 4),
        round(percentage_improvement, 2)
    ]
})

print("\n" + "=" * 70)
print("FINAL KPI OUTPUT")
print("=" * 70)

display(final_kpis)


# ============================================================
# 22. EXPORT PYTHON ANALYSIS RESULTS
# ============================================================

output_file = "/content/Matching_Ranking_Python_Analysis.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    final_kpis.to_excel(
        writer,
        sheet_name="Final KPIs",
        index=False
    )

    category_analysis.to_excel(
        writer,
        sheet_name="Category Analysis",
        index=False
    )

    evaluation_analysis.to_excel(
        writer,
        sheet_name="Evaluation Analysis",
        index=False
    )

    relevance_analysis.to_excel(
        writer,
        sheet_name="Relevance Analysis",
        index=False
    )

    shortlist_analysis.to_excel(
        writer,
        sheet_name="Shortlist Analysis",
        index=False
    )

    hiring_analysis.to_excel(
        writer,
        sheet_name="Hiring Analysis",
        index=False
    )

    ranker_category.to_excel(
        writer,
        sheet_name="Ranker Comparison",
        index=False
    )

    rank_analysis.to_excel(
        writer,
        sheet_name="Rank Position",
        index=False
    )

    strongest_relationships.to_excel(
        writer,
        sheet_name="Correlations",
        index=False
    )

    top_matches.to_excel(
        writer,
        sheet_name="Top Matches",
        index=False
    )

print("\n" + "=" * 70)
print("PYTHON ANALYSIS COMPLETED")
print("=" * 70)

print("\nOutput Excel:")
print(output_file)

print("\nNo charts were created in Python.")
print("Charts and Dashboard will be created in Tableau.")
```
